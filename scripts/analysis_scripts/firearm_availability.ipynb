```python
import pandas as pd
import numpy as np
data_dir = '/home/andrew/abm_violence/data/'
```

```python
# data from https://www.atf.gov/firearms/listing-federal-firearms-licensees/complete
# from 2024 jan - not sure if this is cumulative or just january: it's called "complete listing" so am assuming cumulative
firearm_licenses = pd.read_csv(f"{data_dir}/0124-ffl-list-complete.txt", sep="\t")
major_licenses = firearm_licenses[firearm_licenses["LIC_TYPE"] == 1 | 2 | 7]
major_licenses = major_licenses.drop(columns=["LIC_REGN", 
                                              "LIC_DIST",
                                              "LIC_CNTY",
                                              "LIC_XPRDTE", 
                                              "LIC_SEQN", 
                                              "PREMISE_STREET", 
                                              "MAIL_STREET", 
                                              "MAIL_CITY", 
                                              "MAIL_STATE", 
                                            #   "MAIL_ZIP_CODE", 
                                              "VOICE_PHONE"])
zips = pd.read_csv(f"{data_dir}/2013_us_zips.csv")
```

```python
major_licenses = pd.merge(major_licenses, zips, left_on="PREMISE_ZIP_CODE", right_on="ZIP", how="left")
major_licenses.dropna(inplace=True, subset=["PREMISE_ZIP_CODE", "ZIP"])
major_licenses
```
