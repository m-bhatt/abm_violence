```python
import pandas as pd
import numpy as np
```

```python
# data from https://www.atf.gov/firearms/listing-federal-firearms-licensees/complete
# from 2024 jan - not sure if this is cumulative or just january: it's called "complete listing" so am assuming cumulative
firearm_licenses = pd.read_csv("data/0124-ffl-list-complete.txt", sep="\t")
major_licenses = firearm_licenses[firearm_licenses["LIC_TYPE"] == 1 | 2 | 7]
major_licenses = major_licenses.drop(columns=["LIC_REGN", 
                                              "LIC_DIST",
                                              "LIC_CNTY",
                                              "LIC_XPRDTE", 
                                              "LIC_SEQN", 
                                              "PREMISE_STREET", 
                                              "MAIL_STREET", 
                                              "MAIL_CITY", 
                                              "MAIL_STATE", 
                                            #   "MAIL_ZIP_CODE", 
                                              "VOICE_PHONE"])
zips = pd.read_csv("data/2013_us_zips.csv")
```

```python
set(firearm_licenses['LIC_CNTY'])
```

```python
major_licenses['MAIL_ZIP_CODE'].isna().sum()  # check for missing zip codes
```

```python
known_zips = zips['ZIP'].unique()
unknown_zips = major_licenses[(~major_licenses['PREMISE_ZIP_CODE'].isin(known_zips)) & (~major_licenses['MAIL_ZIP_CODE'].isin(known_zips))]  # check how many zip codes are known
len(unknown_zips)
```

```python
list(unknown_zips['PREMISE_ZIP_CODE'])
```

```python
len(major_licenses)
```

```python
major_licenses = pd.merge(major_licenses, zips, left_on="PREMISE_ZIP_CODE", right_on="ZIP", how="left")
major_licenses.dropna(inplace=True, subset=["PREMISE_ZIP_CODE", "ZIP"])
major_licenses
```

```python
# major_licenses.head(10)
major_licenses.groupby("PREMISE_CITY").agg("count")["BUSINESS_NAME"].sort_values(ascending=False)[0:10]
```
