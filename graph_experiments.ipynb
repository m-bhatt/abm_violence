```python
# option 1: ergm
# lots of nodes, larger edge weights = closer together = higher density
# option 2: graph with different sized nodes, larger node = higher density 

class Graph():
    
    def __init__(self):
        self.adjacency_list = {}
        
    class Node():
        # what will this represent...? a location?
        id_num = 0
        
        def __init__(self, capacity):
            self.capacity = capacity
            #self.description = description
            
            self.id = self.__class__.id_num # cursed
            self.__class__.id_num += 1
            
    def create_node(self, capacity):
        return self.Node(capacity)
    
    def add_edge(self, node1, node2, weight):
        if node1 not in self.adjacency_list.keys():
            self.adjacency_list[node1] = {}
        if node2 not in self.adjacency_list.keys():
            self.adjacency_list[node2] = {}
        # they should point towards each other so both get an edge
        self.adjacency_list[node1][node2] = weight
        self.adjacency_list[node2][node1] = weight
        
    def get_neighbors(self, node):
        return self.adjacency_list[node]
    
    def print_graph(self, start_node):
        # attempted dfs to print all nodes once
        visited_nodes = set()
        visited_nodes.add(start_node)
        print("saw ", start_node.id)
        neighbors = list(self.get_neighbors(start_node).keys()) # returns a dictionary
        
        while len(neighbors) > 0:
            node = neighbors.pop()
            if node not in visited_nodes:
                print("saw ", node.id)
                
                visited_nodes.add(node)
                new_neighbors = list(self.get_neighbors(node).keys())
                neighbors2 = neighbors.copy()
                for n in new_neighbors:
                    if n not in neighbors:
                        neighbors2.append(n)
                neighbors = neighbors2
        print("end")
```

```python
g = Graph()
n1 = g.create_node(1)
n2 = g.create_node(2)
n3 = g.create_node(1)
n4 = g.create_node(3)
g.add_edge(n1, n2, 5)
g.add_edge(n1, n3, 5)
g.add_edge(n2, n3, 10)
g.add_edge(n3, n4, 12)
print(g.get_neighbors(n1))
g.print_graph(n1)
```
